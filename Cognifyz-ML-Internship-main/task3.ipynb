```python
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv("data.csv")
restaurant_name=input("Enter restaurent name:")
selected=df[df["Restaurant Name"]==restaurant_name]
plt.scatter(df["Longitude"], df["Latitude"], color="lightgray", label="All Restaurant")
plt.scatter(
    selected["Longitude"],
    selected["Latitude"],
    marker="X",
    color="red",
    s=100,
    label=restaurant_name
)
for i in range(len(selected)):
    plt.text(
        selected.iloc[i]["Longitude"],
        selected.iloc[i]["Latitude"],
        restaurant_name,
        fontsize=10
    )
plt.scatter(df["Longitude"], df["Latitude"])
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Restaurant Locations")
plt.show()
```
